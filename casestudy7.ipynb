TASK2

```python
import pandas as pd
import os

folder_path = "/content/drive/MyDrive/casestudy7/"
all_files = os.listdir(folder_path) if os.path.exists(folder_path) else []

# Look for a file with 'gdp' in its name dynamically
gdp_file = next((f for f in all_files if "gdp" in f.lower()), None)

if gdp_file:
    full_path = os.path.join(folder_path, gdp_file)
    print(f"Found GDP file: {gdp_file}")
    if gdp_file.endswith(".csv"):
        gdp = pd.read_csv(full_path)
    elif gdp_file.endswith(".json"):
        gdp = pd.read_json(full_path)
    elif gdp_file.endswith(".xml"):
        gdp = pd.read_xml(full_path)
    else:
        gdp = pd.read_excel(full_path)

    # Standardize column to 'country'
    if 'name' in gdp.columns:
        gdp.rename(columns={'name': 'country'}, inplace=True)
    elif 'Country' in gdp.columns:
        gdp.rename(columns={'Country': 'country'}, inplace=True)

    print("\n--- GDP Data Preview ---")
    print(gdp.head())
    print("Columns:", gdp.columns.tolist())
else:
    print("No specific 'gdp' file found in folder. Available files are:", all_files)
```

```python
import pandas as pd

internet_users = pd.read_json("/content/drive/MyDrive/casestudy7/internet_users.json")

print(internet_users.head())
print(internet_users.dtypes)
```

TASK4

```python
import pandas as pd

literacy_rate = pd.read_xml("/content/drive/MyDrive/casestudy7/literacy_rate.xml")

print(literacy_rate.head())
print(literacy_rate.dtypes)
```

PART:B
TASK5

```python
datasets = {
    "Population": population,
    "GDP": gdp,
    "Internet Users": internet_users,
    "Literacy Rate": literacy_rate
}

for name, data in datasets.items():
    print("\n========================")
    print(name)
    print("========================")
    print("Rows:", data.shape[0])
    print("Columns:", data.shape[1])
    print("Column Names:", data.columns.tolist())
    print("\nData Types:")
    print(data.dtypes)
    print("\nMissing Values:")
    print(data.isnull().sum())
    print("Duplicate Records:", data.duplicated().sum())
```

PART:C

TASK6

```python
gdp.rename(columns={"Country": "country"}, inplace=True)
literacy_rate.rename(columns={"name": "country"}, inplace=True)
```

TASK7


```python
population["population"] = pd.to_numeric(population["population"], errors="coerce")
if "GDP" in gdp.columns:
    gdp["GDP"] = pd.to_numeric(gdp["GDP"])
else:
    print("Warning: 'GDP' column not found in gdp DataFrame. Current columns are:", gdp.columns.tolist())

internet_users["internet_users"] = pd.to_numeric(internet_users["internet_users"])
literacy_rate["literacy_rate"] = pd.to_numeric(literacy_rate["literacy_rate"])
```

TASK8

```python
if 'name' in gdp.columns:
    gdp.rename(columns={'name': 'country'}, inplace=True)

df = population.merge(gdp, on="country") \
               .merge(internet_users, on="country") \
               .merge(literacy_rate, on="country")

print(df.head())
```

TASK9

```python
df["Internet Penetration (%)"] = (
    df["internet_users"] / df["population"] * 100
).round(2)

df["GDP Per Capita"] = (
    df["GDP"] * 1000000 / df["population"]
).round(2)

print(df.head())
```

PART:D

TASK10

```python
import sqlite3
conn = sqlite3.connect("country_statistics.db")
df.to_sql(
    "country_statistics",
    conn,
    if_exists="replace",
    index=False
)
print("Data loaded successfully")
```

TASK11

```python
result = pd.read_sql(
    "SELECT * FROM country_statistics LIMIT 5",
    conn
)
print(result)
```
